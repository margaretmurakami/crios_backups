```python
# plot this, and the SFLUX, at two levels
fig = plt.figure(figsize=(15, 10))

def plot_panel(ax, fld, k, title, units="g/m^2/s"):
    vlev = np.nanmax(np.abs(fld * mymsk))
    cb = ax.pcolormesh(
        get_aste_tracer(fld, nfx, nfy)[k],
        cmap="seismic", vmin=-vlev, vmax=vlev
    )
    ax.pcolormesh(
        get_aste_tracer(landmsk, nfx, nfy)[0],
        cmap="Greys", vmin=0, vmax=2
    )
    ax.set_xlim(140, 180)
    ax.set_ylim(140, 180)
    plt.colorbar(cb, ax=ax, label=units)
    ax.set_title(title)

ax = plt.subplot(331)
plot_panel(ax,SFLUX3D,0,"SFLUX k=0")

ax = plt.subplot(332)
plot_panel(ax,flux_sp_top,0,"sp_top k=0")

ax = plt.subplot(333)
plot_panel(ax,flux_sp_bot,0,"sp_bot k=0")

ax = plt.subplot(334)
plot_panel(ax,zconv_sp,0,"zconv_sp k=0")

ax = plt.subplot(335)
plot_panel(ax,oceSPtnd,0,"oceSPtnd k=0")

ax = plt.subplot(336)
plot_panel(ax,(oceSflux-oceSPflx),0,"oceSflux-oceSPflx k=0")
```
