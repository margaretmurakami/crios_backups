```python
def read_all_2d_diags(dirname, file_name, t2, tsstr, ny, nx):
    meta = parsemeta(os.path.join(dirname, file_name + "." + tsstr[0] + ".meta"))
    fldlist = np.array([f.strip() for f in meta["fldList"]])

    out = {}

    for irec, var in enumerate(fldlist):
        arr, its, meta_rd = rdmds(
            os.path.join(dirname, file_name),
            t2,
            returnmeta=True,
            rec=irec
        )

        if arr.size == ny * nx:
            out[var] = arr.reshape(ny, nx)
        else:
            print(f"Skipping {var}: not 2D, shape={arr.shape}")

    return out
```

### now do TFLUX

```python
# Final TFLUX synthesis: closure, components, and ice/open-water context.
# This mirrors the SFLUX workflow but keeps the heat terms in W/m2.
import importlib
import flux_breakdown_plot_options as fbpo
importlib.reload(fbpo)
from flux_breakdown_plot_options import *

mask = mygrid["hFacC"][0]
area = mygrid["RAC"] * mask
theta_surface = THETA[0]
Cp = 3994.0

data = combine_diagnostic_dicts(budg1, budg2, extra_seaice, extra_exf, state2d)

# Main reconstruction from diags_oceanic_surf_flux.F logic used by tflux_options().
FW_heat = oceFWflx * theta_surface * Cp
TFLUX_recon = surForcT + oceFreez + oceQsw + FW_heat
TFLUX_resid = TFLUX - TFLUX_recon

# Direct closure terms and physically interpretable components.
TFLUX_terms = {
    "TFLUX": TFLUX * mask,
    "surForcT": surForcT * mask,
    "oceFreez": oceFreez * mask,
    "oceQsw": oceQsw * mask,
    "oceFWflx * THETA * Cp": FW_heat * mask,
    "TFLUX - surForcT": (TFLUX - surForcT) * mask,
    "surForcT + oceFreez + oceQsw + oceFWflx*THETA*Cp": TFLUX_recon * mask,
    "TFLUX - (surForcT + oceFreez + oceQsw + oceFWflx*THETA*Cp)": TFLUX_resid * mask,
}

plot_option_grid(TFLUX_terms, units="W/m2", ncols=4, robust_pct=99.0)

# Compact numerical summary.  The area integral is W, since W/m2 is multiplied by cell area.
def _area_mean(field, weights):
    return np.nansum(field * weights) / np.nansum(weights)

def _area_sum(field, weights):
    return np.nansum(field * weights)

def _print_tflux_summary(label, selector):
    weights = np.where(selector, area, np.nan)
    print(f"\n{label}")
    print("name                                      area_mean_Wm2       area_sum_W")
    for name, field in TFLUX_terms.items():
        raw = np.asarray(field, dtype=float)
        # TFLUX_terms are already masked; divide by mask only where finite so means are not double-masked.
        values = np.where(np.isfinite(mask) & (mask != 0), raw / mask, np.nan)
        print(f"{name:40s} {_area_mean(values, weights): .6e} {_area_sum(values, weights): .6e}")

water = np.isfinite(mask) & (mask > 0)
ice_covered = water & np.isfinite(SIarea) & (SIarea > 0.15)
open_water = water & np.isfinite(SIarea) & (SIarea <= 0.15)

_print_tflux_summary("All wet surface cells", water)
_print_tflux_summary("Ice-covered cells, SIarea > 0.15", ice_covered)
_print_tflux_summary("Open-water cells, SIarea <= 0.15", open_water)

# Fractional contribution to the reconstructed area-integrated TFLUX.
components = {
    "surForcT": surForcT,
    "oceFreez": oceFreez,
    "oceQsw": oceQsw,
    "oceFWflx * THETA * Cp": FW_heat,
}
recon_sum = _area_sum(TFLUX_recon, area)
print("\nArea-integrated contribution relative to reconstructed TFLUX")
print("name                                      area_sum_W       fraction")
for name, field in components.items():
    asum = _area_sum(field, area)
    frac = asum / recon_sum if np.isfinite(recon_sum) and recon_sum != 0 else np.nan
    print(f"{name:40s} {asum: .6e} {frac: .6e}")

print("\nInterpretation guide:")
print("1. If the residual is near numerical precision, TFLUX is explained by surForcT, oceFreez, oceQsw, and freshwater heat advection.")
print("2. surForcT is the bulk non-shortwave surface forcing path; oceQsw isolates penetrative shortwave handling.")
print("3. oceFreez highlights heat exchanged through freezing/melting; compare it with SIarea and dSIheff patterns.")
```

```python
# Print the values shown in the salt-budget and theta-budget plots at one grid point.
# Indices are Python zero-based array indices.
iy, ix = 10, 10

# Recreate the plotted dictionaries if this cell is run after a kernel restart.
if "synthesis_terms" not in globals():
    rhoIce = 917.0
    SEAICE_salt0 = 4.0
    sice = np.maximum(0.0, np.minimum(SEAICE_salt0, salt_surface))

    sflux_from_oce_salt = oceSflux - oceSPflx
    sflux_from_surfforc_fw = surForcS + oceFWflx * salt_surface
    fw_context_residual = (SFLUX - surForcS) - oceFWflx * salt_surface

    saltFlux_loaded = (
        SIdHbOCN * sice * rhoIce
        + SIdHbATC * sice * rhoIce
        + SIdHbATO * sice * rhoIce
        + SIdHbFLO * sice * rhoIce
    )
    oceSflux_from_loaded_ice = -saltFlux_loaded

    synthesis_terms = {
        "SFLUX diagnostic": SFLUX * hfC0,
        "salt closure: oceSflux - oceSPflx": sflux_from_oce_salt * hfC0,
        "SFLUX - (oceSflux - oceSPflx)": (SFLUX - sflux_from_oce_salt) * hfC0,
        "surForcS + oceFWflx*SALT": sflux_from_surfforc_fw * hfC0,
        "(SFLUX-surForcS)-oceFWflx*SALT": fw_context_residual * hfC0,
        "oceSflux": oceSflux * hfC0,
        "oceSflux_calc=SIdH * S_ice * rhoI": oceSflux_from_loaded_ice * hfC0,
        "oceSflux - oceSflux_calc": (oceSflux - oceSflux_from_loaded_ice) * hfC0,
        "(oceSflux_calc - oceSPflx)": (oceSflux_from_loaded_ice - oceSPflx) * hfC0,
    }

if "TFLUX_terms" not in globals():
    mask = mygrid["hFacC"][0]
    theta_surface = THETA[0]
    Cp = 3994.0
    FW_heat = oceFWflx * theta_surface * Cp
    TFLUX_recon = surForcT + oceFreez + oceQsw + FW_heat
    TFLUX_resid = TFLUX - TFLUX_recon

    TFLUX_terms = {
        "TFLUX": TFLUX * mask,
        "surForcT": surForcT * mask,
        "oceFreez": oceFreez * mask,
        "oceQsw": oceQsw * mask,
        "oceFWflx * THETA * Cp": FW_heat * mask,
        "TFLUX - surForcT": (TFLUX - surForcT) * mask,
        "surForcT + oceFreez + oceQsw + oceFWflx*THETA*Cp": TFLUX_recon * mask,
        "TFLUX - (surForcT + oceFreez + oceQsw + oceFWflx*THETA*Cp)": TFLUX_resid * mask,
    }


# Available oceFWflx / EmPmR-side freshwater diagnostics.
# seaice_growth.F builds EmPmR with the atmospheric term
#   (EVAP - PRECIP) * (1 - AREApreTH) - RUNOFF
# and DIAGS_OCEANIC_SURF_FLUX writes oceFWflx = -EmPmR, so the positive-down
# atmospheric pieces are -EVAP, +PRECIP, and +RUNOFF.  AREApreTH is not in this
# cleanup notebook, so SIarea is used here as the closest available proxy.
if "exf" not in globals():
    exf = read_all_2d_diags(direxf, "exf_zflux_set1", t2, tsstr, ny, nx)

EXFevap = exf.get("EXFevap")
EXFpreci = exf.get("EXFpreci")
EXFroff = exf.get("EXFroff")
rhoFresh = 1000.0
open_water = 1.0 - SIarea

FW_E_down = -EXFevap * open_water * rhoFresh
FW_P_down = EXFpreci * open_water * rhoFresh
FW_R_down = EXFroff * rhoFresh
FW_EPR_down = FW_E_down + FW_P_down + FW_R_down
FW_ice_and_residual_down = oceFWflx - FW_EPR_down

freshwater_terms = {
    "oceFWflx": oceFWflx * hfC0,
    "SIarea": SIarea * hfC0,
    "1 - SIarea": open_water * hfC0,
    "-EXFevap * (1 - SIarea) * rhoFresh": FW_E_down * hfC0,
    "EXFpreci * (1 - SIarea) * rhoFresh": FW_P_down * hfC0,
    "EXFroff * rhoFresh": FW_R_down * hfC0,
    "E/P/R contribution": FW_EPR_down * hfC0,
    "oceFWflx - E/P/R contribution": FW_ice_and_residual_down * hfC0,
    "SIatmFW": SIatmFW * hfC0,
    "SIempmr": SIempmr * hfC0,
    "SIfwSubl": SIfwSubl * hfC0,
    "SIrsSubl": SIrsSubl * hfC0,
    "SIacSubl": SIacSubl * hfC0,
}

def _print_point_terms(title, terms, units, iy, ix):
    print(f"{title} at iy={iy}, ix={ix}")
    print(f"units: {units}")
    print("name                                                           value")
    for name, field in terms.items():
        value = np.asarray(field, dtype=float)[iy, ix]
        print(f"{name:60s} {value: .6e}")
    print("")

print(f"Grid point: iy={iy}, ix={ix}")
if "XC" in mygrid and "YC" in mygrid:
    print(f"XC={mygrid['XC'][iy, ix]:.6f}, YC={mygrid['YC'][iy, ix]:.6f}")
print("")

_print_point_terms("Salt-budget plot values", synthesis_terms, "g/m2/s", iy, ix)
_print_point_terms("Theta-budget plot values", TFLUX_terms, "W/m2", iy, ix)
_print_point_terms("oceFWflx / EmPmR freshwater values", freshwater_terms, "kg/m2/s", iy, ix)
```

```python
# Plot the oceFWflx / EmPmR freshwater breakdown.
# seaice_growth.F uses (EVAP - PRECIP) * (1 - AREApreTH) - RUNOFF in EmPmR.
# AREApreTH is not available here, so SIarea is used as the closest available proxy.
if "exf" not in globals():
    exf = read_all_2d_diags(direxf, "exf_zflux_set1", t2, tsstr, ny, nx)

EXFevap = exf.get("EXFevap")
EXFpreci = exf.get("EXFpreci")
EXFroff = exf.get("EXFroff")
rhoFresh = 1000.0
mask = mygrid["hFacC"][0]
open_water = 1.0 - SIarea

FW_E_down = -EXFevap * open_water * rhoFresh
FW_P_down = EXFpreci * open_water * rhoFresh
FW_R_down = EXFroff * rhoFresh
FW_EPR_down = FW_E_down + FW_P_down + FW_R_down
FW_ice_and_residual_down = oceFWflx - FW_EPR_down

freshwater_breakdown_terms = {
    "oceFWflx": oceFWflx * mask,
    "SIarea": SIarea * mask,
    "1 - SIarea": open_water * mask,
    "-EXFevap * (1 - SIarea) * rhoFresh": FW_E_down * mask,
    "EXFpreci * (1 - SIarea) * rhoFresh": FW_P_down * mask,
    "EXFroff * rhoFresh": FW_R_down * mask,
    "E/P/R contribution": FW_EPR_down * mask,
    "oceFWflx - E/P/R contribution": FW_ice_and_residual_down * mask,
    "SIatmFW": SIatmFW * mask,
    "SIempmr": SIempmr * mask,
    "SIfwSubl": SIfwSubl * mask,
    "SIrsSubl": SIrsSubl * mask,
    "SIacSubl": SIacSubl * mask,
}

plot_option_grid(freshwater_breakdown_terms, units="kg/m2/s or fraction", ncols=4, robust_pct=99.0)
print_integrals(freshwater_breakdown_terms, area=area, label="oceFWflx / EmPmR freshwater breakdown")
```

```python
# also load SIheff from the snaps

file_name = 'budg2d_snap_set1'
meta_budg2d_snap_set1 = parsemeta(dirIn + file_name + "." + tsstr[0] + ".meta")

# try to read this with rec
fldlist = np.array(meta_budg2d_snap_set1['fldList'])
varnames = np.array(['ETAN','PHIBOT','SIheff','SIhsnow'])
recs = np.array([])
for var in varnames:
    irec = np.where(fldlist == var)
    recs = np.append(recs, irec[0][0])

ETAN = np.full((len(tsstr),ny,nx),np.nan)
PHIBOT = np.full((len(tsstr),ny,nx),np.nan)
SIheff = np.full((len(tsstr),ny,nx),np.nan)
SIhsnow = np.full((len(tsstr),ny,nx),np.nan)

for i in range(len(tsstr)):
    thisETAN,its,meta = rdmds(os.path.join(dirIn, file_name),int(tsstr[i]),returnmeta=True,rec=recs[0])
    thisETAN = thisETAN.reshape(ny,nx)
    ETAN[i] = thisETAN

    thisPHIBOT,its,meta = rdmds(os.path.join(dirIn, file_name),int(tsstr[i]),returnmeta=True,rec=recs[1])
    thisPHIBOT = thisPHIBOT.reshape(ny,nx)
    PHIBOT[i] = thisPHIBOT

    thisSIheff,its,meta = rdmds(os.path.join(dirIn, file_name),int(tsstr[i]),returnmeta=True,rec=recs[2])
    thisSIheff = thisSIheff.reshape(ny,nx)
    SIheff[i] = thisSIheff

    thisSIhsnow,its,meta = rdmds(os.path.join(dirIn, file_name),int(tsstr[i]),returnmeta=True,rec=recs[3])
    thisSIhsnow = thisSIhsnow.reshape(ny,nx)
    SIhsnow[i] = thisSIhsnow

ETAN =  (ETAN[1, :, :] - ETAN[0, :, :]) / dt
PHIBOT = (PHIBOT[1, :, :] - PHIBOT[0, :, :]) / dt
dSIheff =  (SIheff[1, :, :] - SIheff[0, :, :]) / dt
dSIhsnow = (SIhsnow[1, :, :] - SIhsnow[0, :, :]) / dt
```

```python
# SFLUX synthesis using snapshot dSIheff instead of loaded SIdHb* terms.
# This mirrors the earlier final SFLUX synthesis cell, but reconstructs oceSflux from
# dSIheff = (SIheff[t2] - SIheff[t1]) / dt read from budg2d_snap_set1.
# This is not expected to close exactly because snapshot dSIheff can include total ice-thickness
# changes, while oceSflux is tied to thermodynamic saltFlux source terms.
rhoIce = 917.0
SEAICE_salt0 = 4.0
mask = mygrid["hFacC"][0]
area = mygrid["RAC"] * mask
sice = np.maximum(0.0, np.minimum(SEAICE_salt0, salt_surface))

# Read SIheff snapshots at t1 and t2.
file_name = "budg2d_snap_set1"
meta_snap = parsemeta(os.path.join(dirIn, file_name + "." + tsstr[0] + ".meta"))
fldlist = np.array(meta_snap["fldList"])
siheff_rec = int(np.where(fldlist == "SIheff")[0][0])

SIheff_snap = np.full((len(tsstr), ny, nx), np.nan)
for nt, ts in enumerate(tsstr):
    this_SIheff, its, meta = rdmds(
        os.path.join(dirIn, file_name),
        int(ts),
        returnmeta=True,
        rec=siheff_rec,
    )
    SIheff_snap[nt] = this_SIheff.reshape(ny, nx)

dSIheff_snap = (SIheff_snap[1] - SIheff_snap[0]) / dt

# Exact diagnostic closures.
sflux_from_oce_salt = oceSflux - oceSPflx
sflux_from_surfforc_fw = surForcS + oceFWflx * salt_surface
fw_context_residual = (SFLUX - surForcS) - oceFWflx * salt_surface

# Snapshot dSIheff reconstruction.  Sign follows oceSflux = -saltFlux.
saltFlux_from_dSIheff = dSIheff_snap * sice * rhoIce
oceSflux_from_dSIheff = -saltFlux_from_dSIheff
sflux_from_dSIheff = oceSflux_from_dSIheff - oceSPflx

synthesis_dSIheff_terms = {
    "SFLUX diagnostic": SFLUX * mask,
    "salt closure: oceSflux - oceSPflx": sflux_from_oce_salt * mask,
    "SFLUX - (oceSflux - oceSPflx)": (SFLUX - sflux_from_oce_salt) * mask,
    "surForcS + oceFWflx*SALT": sflux_from_surfforc_fw * mask,
    "(SFLUX-surForcS)-oceFWflx*SALT": fw_context_residual * mask,
    "oceSflux": oceSflux * mask,
    "oceSflux_calc=-dSIheff*S_ice*rhoI": oceSflux_from_dSIheff * mask,
    "oceSflux - oceSflux_calc": (oceSflux - oceSflux_from_dSIheff) * mask,
    "oceSflux_calc - oceSPflx": sflux_from_dSIheff * mask,
    "SFLUX - (oceSflux_calc - oceSPflx)": (SFLUX - sflux_from_dSIheff) * mask,
    "dSIheff snapshot": dSIheff_snap * mask,
}

plot_option_grid(synthesis_dSIheff_terms, units="g/m2/s or m/s", ncols=4, robust_pct=99.0)
print_integrals(synthesis_dSIheff_terms, area=area, label="SFLUX synthesis from snapshot dSIheff")
```
