```python
# =========================
# OFFLINE divergence diagnostics
# =========================

# total tendency
ttend = termsT_TS['tend'].copy()
stend = termsS_TS['tend'].copy()
ttend[np.isnan(ttend)] = 0
stend[np.isnan(stend)] = 0

dVdt = get_div(ttend, stend, dT_centers, dS_centers)

# individual offline terms
divADVr = get_div(termsT_TS['ADVr'], termsS_TS['ADVr'], dT_centers, dS_centers)
divADVh = get_div(termsT_TS['ADVh'], termsS_TS['ADVh'], dT_centers, dS_centers)
divDFr  = get_div(termsT_TS['DFrT'], termsS_TS['DFrS'], dT_centers, dS_centers)
divDFh  = get_div(termsT_TS['DFhT'], termsS_TS['DFhS'], dT_centers, dS_centers)
divKPP  = get_div(termsT_TS['KPP'],  termsS_TS['KPP'],  dT_centers, dS_centers)
divsurf = get_div(termsT_TS['surf'], termsS_TS['surf'], dT_centers, dS_centers)

# summed diagnostics
divint = divADVr + divADVh + divDFr + divDFh + divKPP
rhs = divint + divsurf
resid = dVdt - rhs

# scale to Sv / degC / psu
lhs_plot      = dVdt    * 1e-6
advr_plot     = divADVr * 1e-6
advh_plot     = divADVh * 1e-6
dfr_plot      = divDFr  * 1e-6
dfh_plot      = divDFh  * 1e-6
kpp_plot      = divKPP  * 1e-6
internal_plot = divint  * 1e-6
surface_plot  = divsurf * 1e-6
rhs_plot      = rhs     * 1e-6
resid_plot    = resid   * 1e-6

# =========================
# Plot selected terms
# =========================

plot_dict = {
    r'LHS: $\partial V/\partial t$': lhs_plot,
    r'$ADV_r$': advr_plot,
    r'$ADV_h$': advh_plot,
    r'$DF_r$': dfr_plot,
    r'$DF_h$': dfh_plot,
    r'$KPP$': kpp_plot,
    r'Surface Forcing': surface_plot,
}

nplots = len(plot_dict)
ncols = 4
nrows = int(np.ceil(nplots / ncols))

fig, axes = plt.subplots(nrows, ncols, figsize=(4.5 * ncols, 4 * nrows), constrained_layout=True)
axes = np.atleast_1d(axes).flatten()

# shared color scale for plotted physical terms
main_fields = [
    np.nan_to_num(lhs_plot, nan=0.0),
    np.nan_to_num(advr_plot, nan=0.0),
    np.nan_to_num(advh_plot, nan=0.0),
    np.nan_to_num(dfr_plot, nan=0.0),
    np.nan_to_num(dfh_plot, nan=0.0),
    np.nan_to_num(kpp_plot, nan=0.0),
    np.nan_to_num(surface_plot, nan=0.0),
]

vlev_main = 5e-5
# or use automatic scaling:
# vlev_main = np.nanmax(np.abs(np.stack(main_fields)))

if not np.isfinite(vlev_main) or vlev_main == 0:
    vlev_main = 1.0

for ax, (title, field) in zip(axes, plot_dict.items()):
    a = field.copy()
    a[a == 0] = np.nan

    pcm = ax.pcolormesh(
        Sbin3, Tbin3, a.T,
        shading='auto',
        cmap='seismic',
        vmin=-vlev_main/2,
        vmax=vlev_main/2
    )
    ax.set_xlabel('Salinity [psu]')
    ax.set_ylabel('Temperature [°C]')
    ax.set_title(title)
    ax.set_xlim(33, 36)
    ax.set_ylim(-2, 8)

    cbar = fig.colorbar(pcm, ax=ax)
    cbar.set_label('Sv/°C/psu')

# delete unused axes
for ax in axes[nplots:]:
    fig.delaxes(ax)

fig.suptitle(
    r'Offline divergence diagnostics with selected offline terms',
    fontsize=14
)

plt.show()
```

```python
# ==========================================
# OFFLINE - ONLINE difference diagnostics
# ==========================================

# offline fields were previously plotted with .T, so put them into
# the same orientation as the online fields before differencing
lhs_off      = lhs_plot.T
advr_off     = advr_plot.T
advh_off     = advh_plot.T
dfr_off      = dfr_plot.T
dfh_off      = dfh_plot.T
kpp_off      = kpp_plot.T
surface_off  = surface_plot.T
internal_off = internal_plot.T
rhs_off      = rhs_plot.T
closure_off  = resid_plot.T

# online fields: no transpose needed
lhs_on      = lhs * 1e-6
vadv_on     = vadv * 1e-6
hadv_on     = hadv * 1e-6
vdiff_on    = vdiff * 1e-6
hdiff_on    = hdiff * 1e-6
surface_on  = surface * 1e-6
residterm_on = resid_term * 1e-6
rhs_on      = rhs * 1e-6
closure_on  = resid * 1e-6

internal_on = vadv_on + hadv_on + vdiff_on + hdiff_on

# differences in matched orientation
diff_lhs     = lhs_off      - lhs_on
diff_advr    = advr_off     - vadv_on
diff_advh    = advh_off     - hadv_on
diff_dfr     = dfr_off      - vdiff_on
diff_dfh     = dfh_off      - hdiff_on
diff_surface = surface_off  - surface_on
diff_internal= internal_off - internal_on

plot_dict = {
    r'$\Delta$ LHS: offline $-$ online': diff_lhs,
    r'$\Delta ADV_r$: offline $-$ online $ADV_z$': diff_advr,
    r'$\Delta ADV_h$: offline $-$ online $ADV_h$': diff_advh,
    r'$\Delta DF_r$: offline $-$ online $DIF_z$': diff_dfr,
    r'$\Delta DF_h$: offline $-$ online $DIF_h$': diff_dfh,
    r'$\Delta$ Surface: offline $-$ online': diff_surface,
    r'$\Delta$ Internal: offline $-$ online': diff_internal,
}

nplots = len(plot_dict)
ncols = 4
nrows = int(np.ceil(nplots / ncols))

fig, axes = plt.subplots(
    nrows, ncols,
    figsize=(4.8 * ncols, 4.2 * nrows),
    constrained_layout=True
)
axes = np.atleast_1d(axes).flatten()

diff_fields = [np.nan_to_num(f, nan=0.0) for f in plot_dict.values()]
vlev_diff = np.nanmax(np.abs(np.stack(diff_fields)))
if not np.isfinite(vlev_diff) or vlev_diff == 0:
    vlev_diff = 1.0

# optional fixed scale
# vlev_diff = 5e-5

for ax, (title, field) in zip(axes, plot_dict.items()):
    a = field.copy()
    a[a == 0] = np.nan

    pcm = ax.pcolormesh(
        Sbin3, Tbin3, a,
        shading='auto',
        cmap='seismic',
        vmin=-vlev_diff/8,
        vmax=vlev_diff/8
    )
    ax.set_xlabel('Salinity [psu]')
    ax.set_ylabel('Temperature [°C]')
    ax.set_title(title)
    ax.set_xlim(33, 36)
    ax.set_ylim(-2, 8)

    cbar = fig.colorbar(pcm, ax=ax)
    cbar.set_label(r'$\Delta$ (Sv/°C/psu)')

for ax in axes[nplots:]:
    fig.delaxes(ax)

fig.suptitle(
    r'Offline $-$ online divergence diagnostics in T-S space',
    fontsize=14
)

plt.savefig("divJonlinevsoffline.png",dpi=300)
plt.show()
```

```python
# what we need to do to achieve an accurate comparison is to be able to normalize these by the bin width at the cell
# we would use THETA and SALT to bin termsT3D and termsS3D, and normalize by the index in binwidthT1 and binwidthS1

def _bincount_sum_with_nan(idx, vals, nout):
    """
    NaN-aware per-bin sum:
    if *all* entries in a bin are NaN, that bin returns NaN;
    otherwise NaNs are ignored and finite values are summed.
    """
    # track counts of non-nan contributions
    finite = np.isfinite(vals)
    sums   = np.bincount(idx[finite], vals[finite], minlength=nout).astype(float)
    counts = np.bincount(idx[finite], None, minlength=nout).astype(float)
    out = sums
    out[counts == 0] = np.nan
    return out
def normalize_and_bin_terms(
    terms3D, tracer3D, binmid, binwidth1
):
    tracer_flat = tracer3D.ravel(order="C")
    nbinm1 = len(binwidth1)

    bin_idx = np.digitize(tracer_flat, binmid, right=False) - 1
    valid = (bin_idx >= 0) & (bin_idx < nbinm1) & np.isfinite(tracer_flat)

    denom_flat = np.full(tracer_flat.shape, np.nan, dtype=float)
    denom_flat[valid] = binwidth1[bin_idx[valid]]

    bin_idx_out = np.full(tracer_flat.shape, -1, dtype=int)
    bin_idx_out[valid] = bin_idx[valid]
    bin_idx_3d = bin_idx_out.reshape(tracer3D.shape, order="C")

    terms_norm_3d = {}
    terms_binned_1d = {}

    for key, arr in terms3D.items():
        arr_flat = np.ravel(arr, order="C").astype(float)

        norm_flat = np.full(arr_flat.shape, np.nan, dtype=float)
        valid_term = valid & np.isfinite(arr_flat)
        norm_flat[valid_term] = arr_flat[valid_term] / denom_flat[valid_term]

        terms_norm_3d[key] = norm_flat.reshape(arr.shape, order="C")
        terms_binned_1d[key] = _bincount_sum_with_nan(
            bin_idx[valid_term],
            norm_flat[valid_term],
            nbinm1
        )

    return terms_norm_3d, terms_binned_1d, bin_idx_3d


# following in m^3/s
termsT3D_norm, termsT_binned, Tbin_idx_3d = normalize_and_bin_terms(
    termsT3D, THETA, binmidT, binwidthT1
)

termsS3D_norm, termsS_binned, Sbin_idx_3d = normalize_and_bin_terms(
    termsS3D, SALT, binmidS, binwidthS1
)
```
