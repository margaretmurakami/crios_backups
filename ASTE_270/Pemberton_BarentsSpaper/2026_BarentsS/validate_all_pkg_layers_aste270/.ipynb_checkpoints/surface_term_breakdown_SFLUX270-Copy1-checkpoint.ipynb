```python
def read_diag_var(dirdiags, file_name, varname, t2, shape):
    meta = parsemeta(os.path.join(dirdiags, file_name + "." + tsstr[0] + ".meta"))
    fldlist = np.array(meta["fldList"])

    irec = np.where(fldlist == varname)[0]
    if len(irec) == 0:
        raise ValueError(f"{varname} not found in {file_name}. Available fields: {fldlist}")

    data, its, meta = rdmds(
        os.path.join(dirdiags, file_name),
        t2,
        returnmeta=True,
        rec=irec[0]
    )

    return data.reshape(shape)


# ============================================================
# budg2d_zflux_set1
# fields: oceFWflx, SFLUX, oceSPflx
# ============================================================

file_name = "budg2d_zflux_set1"

SFLUX = read_diag_var(dirdiags, file_name, "SFLUX", t2, (ny, nx))        # g/m^2/s or psu kg/m^2/s; surface salt flux, positive into ocean
oceFWflx = read_diag_var(dirdiags, file_name, "oceFWflx", t2, (ny, nx)) # kg/m^2/s; net freshwater flux into ocean
oceSPflx = read_diag_var(dirdiags, file_name, "oceSPflx", t2, (ny, nx)) # g/m^2/s or psu kg/m^2/s; salt-plume surface salt flux
SIatmFW = read_diag_var(dirdiags, file_name, "SIatmFW", t2, (ny, nx)) # g/m^2/s or psu kg/m^2/s; salt-plume surface salt flux


# ============================================================
# budg2d_zflux_set2
# fields: oceSflux, SIsnPrcp, SIacSubl
# ============================================================

file_name = "budg2d_zflux_set2"

oceSflux = read_diag_var(dirdiags, file_name, "oceSflux", t2, (ny, nx)) # g/m^2/s or psu kg/m^2/s; net surface salt flux into ocean
SIsnPrcp = read_diag_var(dirdiags, file_name, "SIsnPrcp", t2, (ny, nx)) # kg/m^2/s; snow precipitation onto sea ice
SIacSubl = read_diag_var(dirdiags, file_name, "SIacSubl", t2, (ny, nx)) # kg/m^2/s; accumulated sublimation from sea ice


# ============================================================
# state_2d_set1
# fields: SIarea, SIheff, SIhsnow
# ============================================================

file_name = "state_2d_set1"

SIarea = read_diag_var(dirstate, file_name, "SIarea", t2, (ny, nx))   # 1; sea-ice fractional area
SIheff = read_diag_var(dirstate, file_name, "SIheff", t2, (ny, nx))   # m; effective sea-ice thickness
SIhsnow = read_diag_var(dirstate, file_name, "SIhsnow", t2, (ny, nx)) # m; effective snow thickness on sea ice


# ============================================================
# state_3d_set1
# field: SALT
# take surface level
# ============================================================

file_name = "state_3d_set1"

SALT = read_diag_var(dirstate, file_name, "SALT", t2, (nz, ny, nx)) # psu or g/kg; salinity
SALT_surface = SALT[0, :, :]                                      # psu or g/kg; surface salinity


# ============================================================
# exf_zflux_set1
# fields: EXFpreci, EXFevap, EXFroff, EXFempmr
# ============================================================

file_name = "exf_zflux_set1"

EXFpreci = read_diag_var(dirdiags, file_name, "EXFpreci", t2, (ny, nx)) # m/s; precipitation
EXFevap = read_diag_var(dirdiags, file_name, "EXFevap", t2, (ny, nx))   # m/s; evaporation
EXFroff = read_diag_var(dirdiags, file_name, "EXFroff", t2, (ny, nx))   # m/s; runoff
EXFempmr = read_diag_var(dirdiags, file_name, "EXFempmr", t2, (ny, nx)) # m/s; evap - precip - runoff
```

```python
def print_surface_budget_point(iy, ix, rhoConst=1029.0, Sref=35.0):
    print(f"\nGrid point iy={iy}, ix={ix}")
    print("=" * 70)

    rows = [
        ("SALT_surface", SALT_surface[iy, ix], "psu or g/kg", "surface salinity"),
        ("SIarea", SIarea[iy, ix], "1", "sea-ice fraction"),
        ("SIheff", SIheff[iy, ix], "m", "effective sea-ice thickness"),
        ("SIhsnow", SIhsnow[iy, ix], "m", "snow thickness on sea ice"),
        ("EXFpreci", EXFpreci[iy, ix], "m/s", "precipitation"),
        ("EXFevap", EXFevap[iy, ix], "m/s", "evaporation"),
        ("EXFroff", EXFroff[iy, ix], "m/s", "runoff"),
        ("EXFempmr", EXFempmr[iy, ix], "m/s", "evap - precip - runoff"),
        ("oceFWflx", oceFWflx[iy, ix], "kg/m^2/s", "net freshwater flux"),
        ("SIsnPrcp", SIsnPrcp[iy, ix], "kg/m^2/s", "snow precip on sea ice"),
        ("SIacSubl", SIacSubl[iy, ix], "kg/m^2/s", "sea-ice sublimation"),
        ("oceSflux", oceSflux[iy, ix], "salt flux units", "ocean salt flux"),
        ("oceSPflx", oceSPflx[iy, ix], "salt flux units", "salt-plume flux"),
        ("SFLUX", SFLUX[iy, ix], "salt flux units", "total surface salt flux"),
    ]

    for name, value, units, meaning in rows:
        print(f"{name:15s} {value: .8e}  [{units:12s}]  {meaning}")

    print("\nChecks")
    print("-" * 70)

    exf_empmr_calc = EXFevap[iy, ix] - EXFpreci[iy, ix] - EXFroff[iy, ix]
    print(f"EXFevap - EXFpreci - EXFroff = {exf_empmr_calc: .8e}")
    print(f"EXFempmr                     = {EXFempmr[iy, ix]: .8e}")
    print(f"difference                   = {(EXFempmr[iy, ix] - exf_empmr_calc): .8e}")

    print()

    fw_mps = oceFWflx[iy, ix] / rhoConst
    salt_fw = -Sref * oceFWflx[iy, ix]

    print(f"oceFWflx/rhoConst            = {fw_mps: .8e} m/s")
    print(f"-Sref*oceFWflx               = {salt_fw: .8e}")

    print()

    print(f"SFLUX                        = {SFLUX[iy, ix]: .8e}")
    print(f"oceSflux                     = {oceSflux[iy, ix]: .8e}")
    print(f"oceSPflx                     = {oceSPflx[iy, ix]: .8e}")
    print(f"oceSflux + oceSPflx          = {(oceSflux[iy, ix] + oceSPflx[iy, ix]): .8e}")
    print(f"SFLUX - (oceSflux+oceSPflx)  = {(SFLUX[iy, ix] - (oceSflux[iy, ix] + oceSPflx[iy, ix])): .8e}")


print_surface_budget_point(iy, ix)
```

```python
def walk_surface_salt_forcing_point(iy, ix, rhoConst=1029.0, Sref=35.0):
    print(f"\nSurface forcing walk-through at iy={iy}, ix={ix}")
    print("=" * 80)

    # ------------------------------------------------------------
    # 1. EXF atmospheric freshwater pieces
    # ------------------------------------------------------------
    exf_preci = EXFpreci[iy, ix]   # m/s, precipitation
    exf_evap  = EXFevap[iy, ix]    # m/s, evaporation
    exf_roff  = EXFroff[iy, ix]    # m/s, runoff
    exf_empmr = EXFempmr[iy, ix]   # m/s, E - P - R

    exf_empmr_rebuilt = exf_evap - exf_preci - exf_roff

    print("\n[1] EXF freshwater forcing")
    print(f"EXFevap                       = {exf_evap: .8e}  # m/s")
    print(f"EXFpreci                      = {exf_preci: .8e}  # m/s")
    print(f"EXFroff                       = {exf_roff: .8e}  # m/s")
    print(f"EXFevap - EXFpreci - EXFroff  = {exf_empmr_rebuilt: .8e}  # m/s")
    print(f"EXFempmr                      = {exf_empmr: .8e}  # m/s")
    print(f"residual                      = {(exf_empmr - exf_empmr_rebuilt): .8e}")

    # ------------------------------------------------------------
    # 2. Sea-ice / ocean freshwater exchange
    # In your data.diagnostics:
    # "SIempmr is identical to minus oceFWflx"
    # ------------------------------------------------------------
    oce_fwflx = oceFWflx[iy, ix]   # kg/m^2/s, freshwater flux into ocean

    SIempmr_from_oceFWflx = -oce_fwflx
    oceFWflx_m_per_s = oce_fwflx / rhoConst

    print("\n[2] Ocean/sea-ice freshwater flux")
    print(f"oceFWflx                      = {oce_fwflx: .8e}  # kg/m^2/s")
    print(f"-oceFWflx                     = {SIempmr_from_oceFWflx: .8e}  # SIempmr equivalent")
    print(f"oceFWflx / rhoConst           = {oceFWflx_m_per_s: .8e}  # m/s")

    # ------------------------------------------------------------
    # 3. Local surface salinity used for freshwater-to-salt conversion
    # ------------------------------------------------------------
    s_surf = SALT_surface[iy, ix]  # psu or g/kg

    salt_fw_local = -s_surf * oce_fwflx
    salt_fw_ref   = -Sref * oce_fwflx

    print("\n[3] Freshwater-to-salt-flux conversion")
    print(f"SALT_surface                  = {s_surf: .8e}  # psu or g/kg")
    print(f"-SALT_surface * oceFWflx      = {salt_fw_local: .8e}  # salt flux style units")
    print(f"-Sref * oceFWflx              = {salt_fw_ref: .8e}  # using Sref={Sref}")

    # ------------------------------------------------------------
    # 4. Sea-ice state and sea-ice freshwater terms
    # ------------------------------------------------------------
    print("\n[4] Sea-ice state / sea-ice source terms")
    print(f"SIarea                        = {SIarea[iy, ix]: .8e}  # fraction")
    print(f"SIheff                        = {SIheff[iy, ix]: .8e}  # m")
    print(f"SIhsnow                       = {SIhsnow[iy, ix]: .8e}  # m")
    print(f"SIsnPrcp                      = {SIsnPrcp[iy, ix]: .8e}  # kg/m^2/s")
    print(f"SIacSubl                      = {SIacSubl[iy, ix]: .8e}  # kg/m^2/s")

    # ------------------------------------------------------------
    # 5. Actual saved salt-flux diagnostics
    # Do NOT assume these add unless Fortran says so.
    # Just print them as final diagnosed quantities.
    # ------------------------------------------------------------
    print("\n[5] Saved salt-flux diagnostics")
    print(f"SFLUX                         = {SFLUX[iy, ix]: .8e}  # total surface salt flux diagnostic")
    print(f"oceSflux                      = {oceSflux[iy, ix]: .8e}  # ocean surface salt flux diagnostic")
    print(f"oceSPflx                      = {oceSPflx[iy, ix]: .8e}  # salt-plume flux diagnostic")

    print("\n[6] Useful residuals")
    print(f"oceFWflx/rhoConst - EXFempmr  = {(oceFWflx_m_per_s - exf_empmr): .8e}  # m/s")
    print(f"SFLUX - (-SALT*oceFWflx)      = {(SFLUX[iy, ix] - salt_fw_local): .8e}")
    print(f"oceSflux - (-SALT*oceFWflx)   = {(oceSflux[iy, ix] - salt_fw_local): .8e}")


walk_surface_salt_forcing_point(iy, ix)
```

```python
rho_fw = myparms["rhoconstFresh"]
rho_i  = myparms["rhoi"]
Sice0  = myparms["SIsal0"]
```

```python


# Use SIareaPT if available; otherwise fallback to SIarea
A = SIareaPT if "SIareaPT" in globals() else SIarea

# Surface salinity
Ssurf = SALT[0]  # if SALT is [k,j,i]
# Ssurf = SALT[t,0]  # use this if SALT is [time,k,j,i]

# -----------------------------
# 1. Atmospheric freshwater part of PmEpR / oceFWflx
# -----------------------------
# Code equivalent:
# EmPmR_atm = rho_fw * HEFFM * [(E-P)*(1-A) - R]
# PmEpR_atm = -EmPmR_atm
#
# Assuming HEFFM = 1 over active ocean.
FWatm = rho_fw * (
    EXFpreci * (1 - A)
    + EXFroff
    - EXFevap * (1 - A)
)  # kg/m^2/s

FWatm_salt = FWatm * Ssurf  # g/m^2/s

# -----------------------------
# 2. Ice/snow freshwater residual part
# -----------------------------
# oceFWflx = PmEpR = FWatm + FWice
FWice = oceFWflx - FWatm  # kg/m^2/s

FWice_salt = FWice * Ssurf  # g/m^2/s

# -----------------------------
# 3. Explicit sea-ice salt flux
# -----------------------------
# Diagnostics convention:
# oceSflux = -saltFlux
saltFlux = -oceSflux  # g/m^2/s

# Salinity cap used in seaice_growth.F
S_take = np.maximum(0.0, np.minimum(Sice0, Ssurf))

# Equivalent thermodynamic ice-growth rate inferred from oceSflux
# oceSflux = - dHeff_dt * S_take * rho_i
eps = 1e-20
dHeff_dt_from_oceSflux = -oceSflux / (rho_i * np.maximum(S_take, eps))  # m/s

# -----------------------------
# 4. Reconstruct total SFLUX
# -----------------------------
SFLUX_recon_direct = oceSflux + oceFWflx * Ssurf

SFLUX_recon_split = (
    oceSflux
    + FWatm_salt
    + FWice_salt
)

SFLUX_residual_direct = SFLUX - SFLUX_recon_direct
SFLUX_residual_split  = SFLUX - SFLUX_recon_split

# -----------------------------
# 5. Useful diagnostic checks
# -----------------------------
print("Max abs SFLUX residual direct:",
      np.nanmax(np.abs(SFLUX_residual_direct)))

print("Max abs SFLUX residual split:",
      np.nanmax(np.abs(SFLUX_residual_split)))

print("Max abs difference between reconstructions:",
      np.nanmax(np.abs(SFLUX_recon_direct - SFLUX_recon_split)))

# -----------------------------
# 6. Plot components
# -----------------------------
fields = {
    "oceSflux: explicit sea-ice salt flux": oceSflux,
    "FWatm * Ssurf: atmosphere/runoff freshwater salt effect": FWatm_salt,
    "FWice * Ssurf: sea-ice freshwater residual salt effect": FWice_salt,
    "SFLUX reconstructed": SFLUX_recon_split,
    "SFLUX original": SFLUX,
    "SFLUX residual": SFLUX_residual_split,
}

plt.figure(figsize=(14, 10))

for n, (title, fld) in enumerate(fields.items(), start=1):
    ax = plt.subplot(2, 3, n)
    plotfld = get_aste_tracer(fld, nfx, nfy)[0]
    vlev = np.nanmax(np.abs(fld * mymsk))

    cb = ax.pcolormesh(
        plotfld,
        cmap="seismic",
        vmin=-vlev,
        vmax=vlev
    )

    ax.pcolormesh(land_data, cmap="Greys", vmin=0, vmax=4)
    plt.colorbar(cb, ax=ax, label="g/m$^2$/s")
    ax.set_title(title)
    ax.set_xlim(400, 540)
    ax.set_ylim(400, 600)

plt.tight_layout()
plt.show()

# -----------------------------
# 7. Optional: plot freshwater pieces in kg/m^2/s
# -----------------------------
fw_fields = {
    "oceFWflx": oceFWflx,
    "FWatm": FWatm,
    "FWice residual": FWice,
}

plt.figure(figsize=(12, 4))

for n, (title, fld) in enumerate(fw_fields.items(), start=1):
    ax = plt.subplot(1, 3, n)
    plotfld = get_aste_tracer(fld, nfx, nfy)[0]
    vlev = np.nanmax(np.abs(fld * mymsk))

    cb = ax.pcolormesh(
        plotfld,
        cmap="seismic",
        vmin=-vlev,
        vmax=vlev
    )

    ax.pcolormesh(land_data, cmap="Greys", vmin=0, vmax=4)
    plt.colorbar(cb, ax=ax, label="kg/m$^2$/s")
    ax.set_title(title)
    ax.set_xlim(400, 540)
    ax.set_ylim(400, 600)

plt.tight_layout()
plt.show()
```

```python
# One-cell salt/freshwater diagnostic breakdown
# Assumes iy, ix are already defined.

Ssurf = SALT[0]  # psu ~= g/kg

A = SIarea[iy, ix]
rho_fw = myparms["rhoconstFresh"]
ssurf = Ssurf[iy, ix]

# FWatm breakdown
preci_open = EXFpreci[iy, ix] * (1 - A)
evap_open  = EXFevap[iy, ix]  * (1 - A)
roff_term  = EXFroff[iy, ix]

fwatm_preci = rho_fw * preci_open
fwatm_roff  = rho_fw * roff_term
fwatm_evap  = -rho_fw * evap_open

fwatm = fwatm_preci + fwatm_roff + fwatm_evap
fwice = oceFWflx[iy, ix] - fwatm                 # because we don't have the diagnostics

sflux       = SFLUX[iy, ix]
oce_sflux   = oceSflux[iy, ix]
oce_spflx   = oceSPflx[iy, ix]
si_area     = SIarea[iy, ix]
oce_fwflx   = oceFWflx[iy, ix]

# Freshwater-equivalent salt fluxes
fwatm_preci_salt = fwatm_preci * ssurf
fwatm_roff_salt  = fwatm_roff  * ssurf
fwatm_evap_salt  = fwatm_evap  * ssurf
fwatm_salt       = fwatm * ssurf

fwice_salt       = fwice * ssurf
fw_total_salt    = (fwatm + fwice) * ssurf
ocefw_salt       = oce_fwflx * ssurf

# Sea-ice salt-flux identity
sflux_from_oce = oce_sflux - oce_spflx
residual_sflux_identity = sflux - sflux_from_oce

# Diagnostic reconstruction:
# SFLUX = surfaceForcingS*rUnit2mass + PmEpR*Ssurf
# Assuming oceFWflx == PmEpR.
surfaceForcingS_mass_equiv = sflux - ocefw_salt

print("Direct SFLUX identity")
print(f"SFLUX:                       {sflux:.18e} g/m^2/s")
print(f"oceSflux:                    {oce_sflux:.18e} g/m^2/s")
print(f"-oceSPflx:                   {-oce_spflx:.18e} g/m^2/s")
print(f"oceSflux - oceSPflx:         {sflux_from_oce:.18e} g/m^2/s")
print(f"SFLUX - (oceSflux-oceSPflx): {residual_sflux_identity:.18e} g/m^2/s")

print()
print("Sea-ice state")
print(f"SIarea:                      {si_area:.18e}")
print(f"1 - SIarea:                  {(1 - si_area):.18e}")

print()
print("FWatm breakdown")
print("Volume flux terms:")
print(f"EXFpreci*(1-A):              {preci_open:.18e} m/s")
print(f"EXFroff:                     {roff_term:.18e} m/s")
print(f"-EXFevap*(1-A):              {-evap_open:.18e} m/s")
print(f"sum volume flux:             {(preci_open + roff_term - evap_open):.18e} m/s")

print()
print("Mass flux terms:")
print(f"rho_fw*EXFpreci*(1-A):       {fwatm_preci:.18e} kg/m^2/s")
print(f"rho_fw*EXFroff:              {fwatm_roff:.18e} kg/m^2/s")
print(f"-rho_fw*EXFevap*(1-A):       {fwatm_evap:.18e} kg/m^2/s")
print(f"FWatm sum:                   {fwatm:.18e} kg/m^2/s")

print()
print("Salt-equivalent FWatm terms:")
print(f"rho_fw*EXFpreci*(1-A)*Ssurf: {fwatm_preci_salt:.18e} g/m^2/s")
print(f"rho_fw*EXFroff*Ssurf:        {fwatm_roff_salt:.18e} g/m^2/s")
print(f"-rho_fw*EXFevap*(1-A)*Ssurf: {fwatm_evap_salt:.18e} g/m^2/s")
print(f"FWatm*Ssurf sum:             {fwatm_salt:.18e} g/m^2/s")

print()
print("Freshwater fluxes")
print(f"FWatm:                       {fwatm:.18e} kg/m^2/s")
print(f"FWice = oceFWflx - FWatm:    {fwice:.18e} kg/m^2/s")
print(f"FWatm + FWice:               {(fwatm + fwice):.18e} kg/m^2/s")
print(f"oceFWflx:                    {oce_fwflx:.18e} kg/m^2/s")
print(f"oceFWflx - (FWatm+FWice):    {(oce_fwflx - (fwatm + fwice)):.18e} kg/m^2/s")

print()
print("Freshwater-equivalent salt fluxes")
print(f"Ssurf:                       {ssurf:.18e} g/kg")
print(f"FWatm*Ssurf:                 {fwatm_salt:.18e} g/m^2/s")
print(f"FWice*Ssurf:                 {fwice_salt:.18e} g/m^2/s")
print(f"(FWatm+FWice)*Ssurf:         {fw_total_salt:.18e} g/m^2/s")
print(f"oceFWflx*Ssurf:              {ocefw_salt:.18e} g/m^2/s")
print(f"oceFWflx*Ssurf - sum:        {(ocefw_salt - fw_total_salt):.18e} g/m^2/s")

print()
print("Diagnostic reconstruction")
print("Assuming SFLUX = surfaceForcingS*rUnit2mass + PmEpR*Ssurf")
print("and oceFWflx = PmEpR:")
print(f"surfaceForcingS*rUnit2mass = SFLUX - oceFWflx*Ssurf")
print(f"surfaceForcingS mass equiv:  {surfaceForcingS_mass_equiv:.18e} g/m^2/s")

print()
print("Separated reconstruction:")
print(f"surfaceForcingS mass equiv:  {surfaceForcingS_mass_equiv:.18e} g/m^2/s")
print(f"+ FWatm*Ssurf:               {fwatm_salt:.18e} g/m^2/s")
print(f"+ FWice*Ssurf:               {fwice_salt:.18e} g/m^2/s")
print(f"= reconstructed SFLUX:       {(surfaceForcingS_mass_equiv + fwatm_salt + fwice_salt):.18e} g/m^2/s")
print(f"SFLUX:                       {sflux:.18e} g/m^2/s")
print(f"residual:                    {(sflux - (surfaceForcingS_mass_equiv + fwatm_salt + fwice_salt)):.18e} g/m^2/s")

print()
print("Check using lumped oceFWflx:")

# ocefw_salt = ocefwflx * Ssurf
# ocefwflx = fwatm + fwice = (fwatm_preci_salt + fwatm_roff_salt + fwatm_evap_salt) + fwice_salt
print(f"surfaceForcingS_mass_equiv + fwatm_preci_salt + fwatm_roff_salt + fwatm_evap_salt) + fwice_salt = "
      f"{(surfaceForcingS_mass_equiv + (fwatm_preci_salt + fwatm_roff_salt + fwatm_evap_salt) + fwice_salt):.18e} g/m^2/s")
print(f"surfaceForcingS_mass_equiv + oceFWflx*Ssurf = "
      f"{(surfaceForcingS_mass_equiv + ocefw_salt):.18e} g/m^2/s")
print(f"SFLUX:                                      {sflux:.18e} g/m^2/s")


surfaceForcingS_no_plume_mass_equiv = surfaceForcingS_mass_equiv + oce_spflx

print()
print("=== Check again: individual contributions to SFLUX, plume separated ===")

print(f"surfaceForcingS no plume equiv:    {surfaceForcingS_no_plume_mass_equiv:.18e} g/m^2/s")
print(f"- oceSPflx:                        {-oce_spflx:.18e} g/m^2/s")
print(f"+ precip contribution:             {fwatm_preci_salt:.18e} g/m^2/s")
print(f"+ runoff contribution:             {fwatm_roff_salt:.18e} g/m^2/s")
print(f"- evap contribution:               {fwatm_evap_salt:.18e} g/m^2/s")
print(f"+ ice freshwater contribution:     {fwice_salt:.18e} g/m^2/s")

reconstructed_from_terms_plume_split = (
    surfaceForcingS_no_plume_mass_equiv
    - oce_spflx
    + fwatm_preci_salt
    + fwatm_roff_salt
    + fwatm_evap_salt
    + fwice_salt
)

print("                                ----------------------")
print(f"reconstructed SFLUX:              {reconstructed_from_terms_plume_split:.18e} g/m^2/s")
print(f"actual SFLUX:                     {sflux:.18e} g/m^2/s")
print(f"residual:                         {(reconstructed_from_terms_plume_split - sflux):.18e} g/m^2/s")

print()
print("Algebra:")
print("SFLUX [g/m^2/s] = surfaceForcingS_no_plume*rUnit2mass [g/m^2/s]")
print("                - oceSPflx [g/m^2/s]")
print("                + rho_fw*EXFpreci*(1-A)*Ssurf [g/m^2/s]")
print("                + rho_fw*EXFroff*Ssurf [g/m^2/s]")
print("                - rho_fw*EXFevap*(1-A)*Ssurf [g/m^2/s]")
print("                + FWice*Ssurf [g/m^2/s]")
```

```python
# One-cell heat/freshwater diagnostic breakdown for TFLUX
# Assumes iy, ix are already defined.

Tsurf = THETA[0] if "THETA" in globals() else theta[0]  # degC
tsurf = Tsurf[iy, ix]

Cp = myparms["HeatCapacity_Cp"]  # J/kg/K

A = SIarea[iy, ix]
rho_fw = myparms["rhoconstFresh"]

# Freshwater atmospheric breakdown
preci_open = EXFpreci[iy, ix] * (1 - A)
evap_open  = EXFevap[iy, ix]  * (1 - A)
roff_term  = EXFroff[iy, ix]

fwatm_preci = rho_fw * preci_open
fwatm_roff  = rho_fw * roff_term
fwatm_evap  = -rho_fw * evap_open

fwatm = fwatm_preci + fwatm_roff + fwatm_evap
oce_fwflx = oceFWflx[iy, ix]
fwice = oce_fwflx - fwatm   # this again is the contribution from d_HSNW etc, that we do not have from tmpscal1 in seaice_growth.F

# Heat-equivalent freshwater terms
# kg/m^2/s * J/kg/K * K = W/m^2
fwatm_preci_heat = fwatm_preci * Cp * tsurf
fwatm_roff_heat  = fwatm_roff  * Cp * tsurf
fwatm_evap_heat  = fwatm_evap  * Cp * tsurf
fwatm_heat       = fwatm       * Cp * tsurf
fwice_heat       = fwice       * Cp * tsurf
ocefw_heat       = oce_fwflx   * Cp * tsurf

tflux = TFLUX[iy, ix]

# If you have surForcT as a diagnostic, use it directly:
if "surForcT" in globals():
    surfaceForcingT_heat_equiv = surForcT[iy, ix]
else:
    # From code:
    # TFLUX = surfaceForcingT*Cp*rUnit2mass
    #       + adjustColdSST_diag*Cp*rUnit2mass
    #       - Qsw
    #       + PmEpR*Tsurf*Cp
    #
    # Here we infer the combined non-freshwater-addback part:
    surfaceForcingT_heat_equiv = tflux - ocefw_heat

print("=== Direct TFLUX diagnostic reconstruction ===")
print(f"TFLUX:                         {tflux:.18e} W/m^2")
print(f"Tsurf:                         {tsurf:.18e} degC")
print(f"Cp:                            {Cp:.18e} J/kg/K")
print(f"oceFWflx:                      {oce_fwflx:.18e} kg/m^2/s")
print(f"oceFWflx*Tsurf*Cp:             {ocefw_heat:.18e} W/m^2")

print()
print("Assuming diagnostic form:")
print("TFLUX = surfaceForcingT*Cp*rUnit2mass + oceFWflx*Tsurf*Cp")
print(f"surfaceForcingT heat equiv:    {surfaceForcingT_heat_equiv:.18e} W/m^2")
print(f"+ oceFWflx*Tsurf*Cp:           {ocefw_heat:.18e} W/m^2")
print(f"= reconstructed TFLUX:         {(surfaceForcingT_heat_equiv + ocefw_heat):.18e} W/m^2")
print(f"residual:                      {(surfaceForcingT_heat_equiv + ocefw_heat - tflux):.18e} W/m^2")

print()
print("=== FWatm heat breakdown ===")
print(f"SIarea A:                      {A:.18e}")
print(f"1 - A:                         {(1-A):.18e}")

print()
print("Mass flux terms:")
print(f"rho_fw*EXFpreci*(1-A):         {fwatm_preci:.18e} kg/m^2/s")
print(f"rho_fw*EXFroff:                {fwatm_roff:.18e} kg/m^2/s")
print(f"-rho_fw*EXFevap*(1-A):         {fwatm_evap:.18e} kg/m^2/s")
print(f"FWatm sum:                     {fwatm:.18e} kg/m^2/s")
print(f"FWice = oceFWflx - FWatm:      {fwice:.18e} kg/m^2/s")

print()
print("Heat-equivalent freshwater terms:")
print(f"rho_fw*EXFpreci*(1-A)*Cp*T:    {fwatm_preci_heat:.18e} W/m^2")
print(f"rho_fw*EXFroff*Cp*T:           {fwatm_roff_heat:.18e} W/m^2")
print(f"-rho_fw*EXFevap*(1-A)*Cp*T:    {fwatm_evap_heat:.18e} W/m^2")
print(f"FWatm*Cp*Tsurf:                {fwatm_heat:.18e} W/m^2")
print(f"FWice*Cp*Tsurf:                {fwice_heat:.18e} W/m^2")
print(f"oceFWflx*Cp*Tsurf:             {ocefw_heat:.18e} W/m^2")

print()
print("=== Check individual contributions to TFLUX ===")

reconstructed_from_terms = (
    surfaceForcingT_heat_equiv
    + fwatm_preci_heat
    + fwatm_roff_heat
    + fwatm_evap_heat
    + fwice_heat
)

print(f"surfaceForcingT heat equiv:    {surfaceForcingT_heat_equiv:.18e} W/m^2")
print(f"+ precip heat contribution:    {fwatm_preci_heat:.18e} W/m^2")
print(f"+ runoff heat contribution:    {fwatm_roff_heat:.18e} W/m^2")
print(f"+ evap heat contribution:      {fwatm_evap_heat:.18e} W/m^2")
print(f"+ ice freshwater heat contrib: {fwice_heat:.18e} W/m^2")
print("                              ----------------------")
print(f"reconstructed TFLUX:           {reconstructed_from_terms:.18e} W/m^2")
print(f"actual TFLUX:                  {tflux:.18e} W/m^2")
print(f"residual:                      {(reconstructed_from_terms - tflux):.18e} W/m^2")
```

```python
fig = plt.figure(figsize=(20, 12))

Tsurf = THETA[0] if "THETA" in globals() else theta[0]  # map, degC
Cp = myparms["HeatCapacity_Cp"]                         # J/kg/K
A = SIarea
rho_fw = myparms["rhoconstFresh"]

ocefw_heat = oceFWflx * Cp * Tsurf

# If you have surForcT, use it directly.
# Otherwise infer the combined non-freshwater-addback part from TFLUX.
if "surForcT" in globals():
    surfaceForcingT_heat_equiv = surForcT
else:
    surfaceForcingT_heat_equiv = TFLUX - ocefw_heat

# FWatm breakdown
preci_open = EXFpreci * (1 - A)
evap_open  = EXFevap  * (1 - A)
roff_term  = EXFroff

fwatm_preci = rho_fw * preci_open
fwatm_roff  = rho_fw * roff_term
fwatm_evap  = -rho_fw * evap_open

fwatm = fwatm_preci + fwatm_roff + fwatm_evap
fwice = oceFWflx - fwatm

# Heat-equivalent freshwater terms
fwatm_preci_heat = fwatm_preci * Cp * Tsurf
fwatm_roff_heat  = fwatm_roff  * Cp * Tsurf
fwatm_evap_heat  = fwatm_evap  * Cp * Tsurf
fwice_heat       = fwice       * Cp * Tsurf

reconstructed_from_terms_heat = (
    surfaceForcingT_heat_equiv
    + fwatm_preci_heat
    + fwatm_roff_heat
    + fwatm_evap_heat
    + fwice_heat
)

def plot_panel(pos, a, title, label="W/m^2"):
    ax = plt.subplot(pos)
    aplot = get_aste_tracer(a, nfx, nfy)[0]
    vlev = np.nanmax(np.abs(a * mymsk))
    if not np.isfinite(vlev) or vlev == 0:
        vlev = 1.0

    cb = ax.pcolormesh(aplot, cmap="seismic", vmin=-vlev, vmax=vlev)
    ax.pcolormesh(land_data, cmap="Greys", vmin=0, vmax=4)
    plt.colorbar(cb, label=label)
    ax.set_title(title)
    ax.set_xlim(400, 540)
    ax.set_ylim(400, 600)
    return ax

plot_panel(341, surfaceForcingT_heat_equiv,
           "surfaceForcingT heat equiv\nsurForcT or inferred: TFLUX - oceFWflx*Cp*Tsurf")

plot_panel(342, fwatm_preci_heat,
           "rho_fw*EXFpreci*(1-A)*Cp*Tsurf")

plot_panel(343, fwatm_roff_heat,
           "rho_fw*EXFroff*Cp*Tsurf")

plot_panel(344, fwatm_evap_heat,
           "-rho_fw*EXFevap*(1-A)*Cp*Tsurf")

plot_panel(345, fwice_heat,
           "FWice*Cp*Tsurf\nFWice = oceFWflx - FWatm")

plot_panel(346, ocefw_heat,
           "oceFWflx*Cp*Tsurf")

plot_panel(347, reconstructed_from_terms_heat,
           "reconstructed TFLUX")

plot_panel(348, TFLUX,
           "actual TFLUX")

plot_panel(349, reconstructed_from_terms_heat - TFLUX,
           "reconstruction - TFLUX")

plt.tight_layout()
```
